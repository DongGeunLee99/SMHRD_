### opencv의 기본 기능 - 이미지

```python
# 이미지 불러오기
# cv2.imread() : 괄호 내부의 이미지 경로에 접근해서 이미지를 불러오는 방법
img = cv2.imread('./data/pets.jpg')
cv2.imshow("img", img)
cv2.waitKey(0) # 키 입력을 대기 -> 0이란 숫자가 들어있다면 무한 대기
cv2.destroyAllWindows()
```

```python
import matplotlib.pyplot as plt
# matplotlib(python)의 색상 채널 구조 : RGB
# opencv 색상 채널 구조 : BGR
img = cv2.cvtColor ( img , cv2.COLOR_RGB2BGR )

plt.xticks([])
plt.yticks([])
plt.imshow(img)
```

```python
cap = cv2.VideoCapture(0) # 연결된 0번 카메라에서 영상 정보를 불러온다
# cap = cv2.VideoCapture('./data/dogs.mp4') # 동영상을 불러온다

while cap.isOpened(): # 비디오 객체가 잡힌다면 True
    ret, frame = cap.read() # ret : 성공 여부, frame : 데이터

    img_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    flipped = cv2.flip(img_gray, 1)

    if not ret :
        print("동영상 재생 끝 or 읽기 실패")
        break

    # 이미지 읽어오기
    cv2.imshow('video1', flipped)
    cv2.imshow('video2', img_gray)


    # 입력키 설정하기
    key = cv2.waitKey(33) # 동영상이 실행되는 시간에 대해 프레임을 제어
    
    # if key == 49: # 아스키코드 값
    if key == ord("q"):
        print("사용자에 의해 영상이 종료")
        break

cap.release() # 카메라 객체 닫아주기
cv2.destroyAllWindows() # 전체 창 닫기
```

```python
img = cv2.imread('./data/cute.png')
faces, conf = cv.detect_face(img) # 얼굴을 탐지하는 기능을 구현한 함수
print(faces) # 얼굴 탐지 픽셀 좌표
print(conf) # 인식 정확도(확률)

cv2.imshow('cute', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 바운딩박스
import numpy as np
img_bg = np.zeros((480, 640, 3)) # 해당 크기의 배열을 생성 후 0으로 채우기
line_color = [0, 255, 0]
line_thick = 4

cv2.rectangle(img_bg, (0, 240), (240, 480) ,line_color, line_thick) # : 이미지에 직사각형을 그려주는 함수
cv2.rectangle(img_bg, (360, 240), (600, 480) ,line_color, cv2.FILLED) # : 이미지에 직사각형을 그려주는 함수


cv2.imshow('img_bg', img_bg)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 0번째 사람의 얼굴 영역에 사각형을 그려보기

img = cv2.imread('./data/cute.png')
faces, conf = cv.detect_face(img) # 얼굴을 탐지하는 기능을 구현한 함수

line_color = [0, 255, 255]
line_thick = 4

for i in range(len(faces)):
    # cv2.putText(이미지, 텍스트, 시작위치, 글꼴, 크기, 색, 두깨)
    cv2.putText(img, f'{conf[i]}', (faces[i][0], faces[i][1]-10),
                cv2.FONT_HERSHEY_PLAIN, 1.5, line_color, line_thick)
    cv2.rectangle(img, (faces[i][0], faces[i][1]), (faces[i][2], faces[i][3]) ,line_color, line_thick) # : 이미지에 직사각형을 그려주는 함수

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 0번째 사람의 얼굴 영역에 사각형을 그려보기 - 얼굴 영역과 인식확률을 하나로 묶어서 사용
img = cv2.VideoCapture('./data/dogs.mp4')
# img = cv2.imread('./data/cute.png')
faces, conf = cv.detect_face(img)

line_color = [0, 255, 255]
line_thick = 4

for (x, y, x2, y2), c in zip(faces, conf):
    cv2.putText(img, f'{c:.2f}', (x, y-10),
                cv2.FONT_HERSHEY_PLAIN, 1.5, [255, 255, 255], line_thick)
    cv2.rectangle(img, (x, y), (x2, y2) ,line_color, line_thick)

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 카메라 영상에서 얼굴 인식하기

```python
import cv2
import cvlib as cv

cap = cv2.VideoCapture(0)
color = [0, 255, 255]
line_th = 4
cnt = 0


while 1: # 비디오 객체가 잡힌다면 True
    try:
        ret, frame = cap.read()
        flip_frame = cv2.flip(frame, 1)
        if not ret:
            print("동영상 재생 끝 or 영상 부르기 실패")
            break


        # 얼굴탐지
        faces, conf = cv.detect_face(flip_frame)

        for (x, y, x2, y2), c in zip(faces, conf):
            cv2.putText(flip_frame, f'{c:.2f}', (x, y-10),
                        cv2.FONT_HERSHEY_PLAIN, 1.5, [255, 255, 255], line_th)
            cv2.rectangle(flip_frame, (x, y), (x2, y2) ,color, line_th)

        cv2.imshow('frame', flip_frame)
        
        key = cv2.waitKey(33)
        if key == ord("q"):
            print("사용자에 의해 영상이 종료")
            cv2.destroyAllWindows()
            break

        if key == ord("w"):
            cnt += 1
            print(f'./data/face_dt{cnt}.png 사진 저장')
            cv2.imwrite('./data/face_dt{cnt}.png', flip_frame, params=[cv2.IMWRITE_PNG_COMPRESSION, 0])
            # cv2.imwrite(저장경로/파일명.확장자, 프레임, params=[이미지 압축, 압축 강도(정수)])

        

    except:
        print("except >> 동영상 재생 끝 or 영상 부르기 실패")
        cap.release() # 카메라 연결 해제
        cv2.distroyAllWindows()

cap.release()

```

```python
# 이미지 출력해보기
import matplotlib.pyplot as plt
for i in range(1, 3):
    img = cv2.imread(f'./data/face_dt{i}.png')
    img = cv2.cvtColor(img, cv2.color_BGR2RGB)
    plt.subplot(1, 2, i)
    plt.imshow(img)
```
