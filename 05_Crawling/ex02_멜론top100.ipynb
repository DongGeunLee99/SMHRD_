- 멜론 페이지에서 노래제목, 가수명을 수집
- 다수의 데이터를 수집하고, 표 형태로 저장해보자

```python
import requests as req
from bs4 import BeautifulSoup as bs
import pandas as pb
```

```python
url = 'https://www.melon.com/chart/index.htm'
res = req.get(url)
print(res)
```

```python
# 멜론페이지에서 컴퓨터(python)로 접근하는 것을 막고 있기 때문에 -> 우회접속
# F12 개발자모드 -> NetWork -> F5 -> index.htm(document타입) -> user-agent를 dictionary형태로 선언
# req.get(url,headers=h)
h = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/132.0.0.0 Safari/537.36'}
res = req.get(url,headers=h)
```

```python
res
```

```python
soup = bs(res.text, 'lxml')
soup
```

### 데이터수집
- 노래제목
- 가수(아티스트)

```python
# 선택자 이름 안에 공백이 있으면 자손 선택자를 의미 ->
# 복수의 클래스를 선택하기 위해선 띄어쓰기를 .으로 대체해야 한다.
# 개발자모드로 해당 태그 마우스 올리면 화면에 나오는 형식을 따라하면 된다. 

song = soup.select('div.ellipsis.rank01')
print(song[0].text)

# print(len(song))
# song[0].text
```

```python
singer = soup.select('span.checkEllipsis')
print(singer[0].text)
```

```python
# 데이터 전처리
# 노래 제목 앞, 뒤에 있는 개행문자를 제거해보자
print(song[0].text.strip())
```

```python
# 반복문 실습
# 노래 제목, 가수명, 순위
song_list = []
singer_list = []
rank = []

for i in range(100):
    song_list.append(song[i].text.strip())
    singer_list.append(singer[i].text.strip())
    rank.append(i+1)
```

```python
dic_melon = {'노래제목' : song_list, '아티스트' :singer_list, '순위' :rank}
melon = pb.DataFrame(dic_melon)
```

```python
melon.set_index('순위', inplace=True)
```
