Fast API 프레임워크 만들기

```python
# 1. uvicorn 안에서 작동시킬 객체 생성
# Spring Boot RestController 역할
app = FastAPI()


# CORS (동일출처정책) -> Spring boot에서 접근을 허용하게 설정
app.add_middleware(
    CORSMiddleware,
    # 요청을 허용할 url 리스트 형태로 작성
    allow_origins = ["http://localhost:8088/web/", "http://localhost:8088/"],
    # 어떤 요청 방식을 허용할건지
    allow_methods = ["*"],
    # 어떤 요청 헤더를 허용할건지
    allow_headers = ["*"],
    allow_credentials = True
)

# 2. 요청을 받아올 url mapping
@app.get("/") # app.get("/url값") / app.post("/url값")
def test( test : str): # url값으로 요청이 들어왔을 때, 실행할 함수 작성
    # 받아올 데이터 키 값은 query String 형식으로 받아옴
    # query String이란 url?key=value
    # 요청 시 수집할 데이터가 있다면, 매개변수로 받아오면 된다.
    # (받아올 데이터 키값 : 자료형,)            여기서 ,는 여러개의 데이터를 받아올 때
    test += "fast결과값"
    return {"test" : test}



@app.post("/files")
def getfile(file : UploadFile=File(...)):

    return {"file_name" : file.filename}
```
